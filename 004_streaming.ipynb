```python
# Create an API client
from anthropic import Anthropic


client = Anthropic()
model = "claude-sonnet-4-5"
```

```python
from IPython.display import Markdown, display

def show(text):
    display(Markdown(text))

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    # anthropic 1.x removed `temperature` from the create() signature, so passing it
    # as a normal keyword raises TypeError. extra_body puts it back on the request body.
    message = client.messages.create(**params, extra_body={"temperature": temperature})
    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "Write a 1 sentence description og a fake database")

stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True
)
for event in stream:
    print(event)
```

```python
messages = []

add_user_message(messages, "Write a 1 sentence description of a fake database")

with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        print(text, end="")
```
