````python
# Create an API client
from anthropic import Anthropic


client = Anthropic()
model = "claude-sonnet-5"
````

````python
from IPython.display import Markdown, display

def show(text):
    display(Markdown(text))

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)



def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)
    # The model may return a thinking block first, so pick the first text block
    return next(block.text for block in message.content if block.type == "text")
````

````python
messages = []

# Prefill is not supported by this model, so ask for the format in the prompt instead
add_user_message(messages, "Generate a very short event bridge rule as json. Respond only with the JSON inside a ```json code block.")

text = chat(messages, stop_sequences=["```\n"])
text
````

````python
import json

# Remove the ```json fence that the model puts in front of the JSON
json.loads(text.strip().removeprefix("```json").strip())
````
